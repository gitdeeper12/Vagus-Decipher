# AWIE Demo - Adaptive Wavelet Isolation Engine

Isolating immune-afferent spike trains from cardiac and respiratory noise

```python
import sys
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
from vagus_decipher.awie import WaveletDecomposer, SpatiotemporalBeamformer, SpikeDetector
```

```python
# Create synthetic ENG signal with spikes
fs = 30000
duration = 2.0
t = np.arange(0, duration, 1/fs)

# Add synthetic spikes
signal = np.random.randn(len(t)) * 0.1
spike_positions = np.random.randint(0, len(t), 20)
for pos in spike_positions:
    if pos + 10 < len(t):
        signal[pos:pos+10] += 1.0

# Add cardiac noise (60 Hz)
cardiac = 0.5 * np.sin(2 * np.pi * 60 * t)
signal += cardiac
```

```python
# Apply AWIE
decomposer = WaveletDecomposer(fs=fs)
isolated = decomposer.process(signal)

# Plot comparison
fig, axes = plt.subplots(2, 1, figsize=(12, 6))
axes[0].plot(t[:5000], signal[:5000])
axes[0].set_title('Raw ENG Signal (with cardiac noise)')
axes[0].set_ylabel('Amplitude')

axes[1].plot(t[:5000], isolated[:5000])
axes[1].set_title('Isolated Immune-Afferent Signal (AWIE)')
axes[1].set_xlabel('Time (s)')
axes[1].set_ylabel('Amplitude')
plt.tight_layout()
plt.show()
```
