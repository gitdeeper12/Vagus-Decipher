# NISSD Demo - Neuro-Immune State-Space Decoder

Mapping firing rate to immunological state vector

```python
import sys
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
from vagus_decipher.nissd import NeuroImmuneStateSpaceModel, UnscentedKalmanFilter, JacobianConstraint
```

```python
# Initialize NISSD
model = NeuroImmuneStateSpaceModel(state_dim=7)
ukf = UnscentedKalmanFilter(state_dim=7, meas_dim=7)

print("State components:", model.get_state_names())
```

```python
# Simulate inflammatory response
times = np.arange(0, 100, 1)
firing_rates = 5.0 + 0.5 * times  # Increasing firing rate

states = []
S = np.zeros(7)

for t, Lambda in zip(times, firing_rates):
    Lambda_vec = np.array([Lambda] * 7)
    S = model.state_transition(S, Lambda_vec, dt=1.0)
    states.append(S.copy())

states = np.array(states)
```

```python
# Plot cytokine trajectories
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

axes[0].plot(times, firing_rates, 'b-', linewidth=2)
axes[0].set_ylabel('Firing Rate (spikes/s)')
axes[0].set_title('Immune-Afferent Firing Rate')

for i, name in enumerate(['TNF-α', 'IL-6', 'IL-10']):
    axes[1].plot(times, states[:, i], label=name, linewidth=2)

axes[1].set_xlabel('Time (minutes)')
axes[1].set_ylabel('Concentration (pg/mL)')
axes[1].set_title('Cytokine State Estimates')
axes[1].legend()
plt.tight_layout()
plt.show()
```
