```python
import sys
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
from vagus_decipher import VagusDecipherEngine
from vagus_decipher.utils import generate_synthetic_eng
```

```python
# Initialize engine
engine = VagusDecipherEngine(
    interface='implanted_cuff',
    n_contacts=6,
    fs=30000,
    warn_horizon_min=45
)
print("✅ Engine initialized")
```

```python
# Generate synthetic ENG data
duration = 10.0  # seconds
signal = generate_synthetic_eng(duration, fs=30000, spike_rate=5.0)
print(f"Signal shape: {signal.shape}")
print(f"Sampling rate: 30 kHz")
```

```python
# Process signal in chunks
chunk_size = int(0.1 * 30000)  # 100 ms chunks
results = []

for i in range(0, len(signal), chunk_size):
    chunk = signal[i:i+chunk_size]
    if len(chunk) == chunk_size:
        result = engine.process(chunk)
        results.append(result)

print(f"Processed {len(results)} chunks")
```

```python
# Plot ISI over time
isi_values = [r['isi'] for r in results]
times = np.arange(len(isi_values)) * 0.1

plt.figure(figsize=(12, 4))
plt.plot(times, isi_values, 'b-', linewidth=2)
plt.axhline(y=0.35, color='y', linestyle='--', label='Elevated Threshold')
plt.axhline(y=0.55, color='orange', linestyle='--', label='High Threshold')
plt.axhline(y=0.75, color='r', linestyle='--', label='Critical Threshold')
plt.xlabel('Time (seconds)')
plt.ylabel('ISI Score')
plt.title('Inflammatory Storm Index Over Time')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Display final results
final = results[-1]
print(f"\n📊 Final Results:")
print(f"  ISI Score: {final['isi']:.3f}")
print(f"  Alert Level: {final['alert_level']}")
print(f"  Alert Message: {final['alert_message']}")
print(f"  Response Time: {final['response_time_min']} minutes")
print(f"\n🧬 Cytokine State Estimates:")
for key, value in final['state'].items():
    print(f"  {key}: {value:.2f} pg/mL" if 'IL' in key or 'TNF' in key else f"  {key}: {value:.2f}")
```
