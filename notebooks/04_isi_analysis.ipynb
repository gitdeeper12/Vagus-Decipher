# ISI Analysis - Inflammatory Storm Index Predictor

Real-time risk scoring with 30-60 minute advance warning

```python
import sys
sys.path.append('..')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from vagus_decipher.isi import InflammatoryStormIndex, ClinicalThresholds
```

```python
# Load LPS endotoxemia data
df = pd.read_csv('../data/examples/lps_endotoxemia.csv')
print(df.head())
```

```python
# Plot ISI trajectory
plt.figure(figsize=(12, 5))
plt.plot(df['time_min'], df['ISI'], 'b-', linewidth=2, marker='o')

# Add threshold lines
plt.axhline(y=0.35, color='y', linestyle='--', label='Elevated (0.35)')
plt.axhline(y=0.55, color='orange', linestyle='--', label='High (0.55)')
plt.axhline(y=0.75, color='r', linestyle='--', label='Critical (0.75)')

# Mark alert time
alert_idx = df[df['ISI'] > 0.65].index[0] if any(df['ISI'] > 0.65) else None
if alert_idx:
    plt.axvline(x=df.loc[alert_idx, 'time_min'], color='r', linestyle=':', label=f"Alert at {df.loc[alert_idx, 'time_min']} min")

plt.xlabel('Time (minutes)')
plt.ylabel('ISI Score')
plt.title('Inflammatory Storm Index - LPS Endotoxemia (M1)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Alert level distribution
levels = df['alert_level'].value_counts()
colors = {'Low': 'green', 'Elevated': 'yellow', 'High': 'orange', 'Critical': 'red'}
plt.bar(levels.index, levels.values, color=[colors.get(l, 'gray') for l in levels.index])
plt.xlabel('Alert Level')
plt.ylabel('Count')
plt.title('Alert Level Distribution')
plt.show()
```
